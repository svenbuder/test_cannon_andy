```python
import numpy as np
import scipy.optimize as op
import matplotlib.pyplot as plt
```

```python
def f(x, a, b, c):
    return a * np.sin(x) + b * np.cos(x) + c * np.tan(x)
```

```python
f(1.0, 1.0, 1.0, 1.0)
```

```python
xpts = list(range(0, 100))
data = [f(x, 2.0, 3.0, 1.88) for x in xpts] + np.random.random((100, ))
```

```python
ax = plt.gca()
ax.plot(range(0, 100), data)
```

```python
def residuals(x):
    a, b, c = x[0], x[1], x[2]
    return np.vectorize(f)(xpts, a, b, c) - data
residuals((2, 3, 1.88))
```

```python
fit_result = op.least_squares(residuals, x0=(2.5, 14.5, 2.5), bounds=(2.5, np.inf))
fit_result
```
